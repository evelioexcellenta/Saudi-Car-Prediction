## Business Understanding

Pasar mobil bekas di Arab Saudi merupakan industri yang terus tumbuh dengan dinamika harga yang kompleks, dipengaruhi oleh berbagai atribut kendaraan seperti merk, tahun pembuatan, ukuran mesin, jenis bahan bakar, jarak tempuh, dan kondisi fitur kendaraan (misalnya Full, Semi Full, atau Standard). Dalam konteks ini, platform jual-beli mobil bekas, showroom, atau bahkan individu yang ingin menjual mobilnya memerlukan pemahaman yang lebih baik tentang bagaimana atribut-atribut tersebut memengaruhi harga pasar mobil bekas.

Tujuan dari analisis ini adalah untuk membantu penjual dan pembeli membuat keputusan yang lebih baik terkait harga yang wajar, serta mengidentifikasi faktor-faktor yang paling signifikan dalam menentukan nilai jual mobil bekas. Selain itu, model prediktif dapat digunakan untuk memperkirakan harga mobil bekas berdasarkan atribut-atribut yang tersedia.
