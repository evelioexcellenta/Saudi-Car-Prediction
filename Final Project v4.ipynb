#### 3.4 Distribusi Fitur Numerik

Selanjutnya kita akan melakukan tes distribusi, 
Mengapa Perlu Melakukan Tes Distribusi Normal dan Menggunakan Shapiro-Wilk Test?

- **Memahami Karakteristik Data:**  
  Tes distribusi normal membantu kita memahami apakah data mengikuti pola distribusi normal (Gaussian). Ini penting karena banyak analisis statistik dan model prediktif mengasumsikan data atau residual yang berdistribusi normal.

- **Validasi Asumsi Statistik:**  
  Beberapa metode statistik (misalnya, uji parametrik atau analisis regresi) mengandalkan asumsi normalitas. Jika data tidak normal, hasil uji statistik mungkin tidak valid atau interpretasinya bisa menyesatkan.

- **Pemilihan Teknik Pemodelan:**  
  Mengetahui distribusi data memungkinkan kita menentukan apakah perlu dilakukan transformasi data (seperti log-transformasi) untuk mencapai normalitas atau memilih model yang tidak terlalu sensitif terhadap distribusi data.

- **Mengapa Shapiro-Wilk Test?**  
  - **Sensitivitas:** Shapiro-Wilk adalah salah satu tes yang paling sensitif untuk mendeteksi penyimpangan dari distribusi normal, terutama pada sampel berukuran kecil hingga sedang.
  - **Kemudahan Implementasi:** Tes ini mudah diimplementasikan dengan library seperti SciPy dan memberikan p-value yang membantu dalam pengambilan keputusan (misalnya, p-value < 0.05 menunjukkan data kemungkinan tidak berdistribusi normal).

Dengan melakukan tes distribusi normal, kita dapat memastikan bahwa asumsi-asumsi penting terpenuhi atau mengetahui kapan harus melakukan penyesuaian agar analisis atau model yang dibangun lebih akurat dan valid.


```python
import plotly.express as px
from scipy.stats import shapiro
import warnings
warnings.simplefilter("ignore")

numeric_cols = ['Price', 'Mileage', 'Year', 'Engine_Size']

for col in numeric_cols:
    print("="*60)
    print(f"Analisis Kolom: {col}")
    
    # 1. Histogram + KDE menggunakan Plotly
    fig = px.histogram(df, x=col, nbins=50, marginal="box", title=f'Distribusi Kolom {col}', opacity=0.7)
    fig.update_layout(bargap=0.2)
    fig.show()
    
    # 2. Shapiro-Wilk Test
    # Pastikan kolom tidak mengandung NaN sebelum dijalankan
    data_non_null = df[col].dropna()
    stat, p_val = shapiro(data_non_null)
    
    print(f"Shapiro-Wilk Test untuk kolom {col}:")
    print(f"  Statistik = {stat:.4f}, p-value = {p_val:.4f}")
    if p_val < 0.05:
        print("  => Data kemungkinan TIDAK terdistribusi normal (Reject H0).")
    else:
        print("  => Data berpotensi terdistribusi normal (Fail to reject H0).")
    print("="*60)
    print("\n")
```

**Important Note** : Karena datanya tidak terdistribusi normal maka kita akan menggunakan metode IQR untuk mendeteksi outliers.

### Menggunakan Metode IQR untuk Deteksi Outliers

- **Tidak Bergantung pada Asumsi Normalitas:**  
  IQR merupakan metode non-parametrik yang tidak mengasumsikan data berdistribusi normal, sehingga cocok untuk dataset yang tidak terdistribusi normal.

- **Robust terhadap Outliers:**  
  Dengan memfokuskan analisis pada rentang kuartil (Q1 sampai Q3), IQR mengabaikan nilai ekstrem sehingga lebih tahan terhadap pengaruh outlier dibandingkan dengan metode berbasis mean dan standar deviasi.

- **Simplicity dan Efektivitas:**  
  IQR mudah dihitung dan diinterpretasikan dengan menggunakan aturan 1.5 * IQR untuk menentukan batas bawah dan atas, membuatnya menjadi pendekatan yang efektif untuk mendeteksi nilai-nilai yang menyimpang secara signifikan.

- **Fokus pada Data Tengah:**  
  Metode ini menangkap informasi dari 50% data tengah (antara Q1 dan Q3), yang memberikan gambaran yang lebih stabil tentang sebaran data dan membantu mengidentifikasi outliers dengan lebih tepat.


```python
numeric_columns = ['Year', 'Engine_Size', 'Mileage', 'Price']

for col in numeric_columns:
    # 1. Menghitung Q1, Q3, dan IQR
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    # 2. Menentukan batas bawah dan batas atas
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # 3. Filter baris-baris yang termasuk outlier
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    # 4. Menampilkan jumlah outlier dan persentase relatif terhadap total baris
    total_rows = len(df)
    num_outliers = len(outliers)
    percentage_outliers = (num_outliers / total_rows) * 100

    print(f"Kolom: {col}")
    print(f"Jumlah Outlier: {num_outliers} dari {total_rows} baris data ({percentage_outliers:.2f}%).")
    print(f"Batas Bawah = {lower_bound}, Batas Atas = {upper_bound}")
    
    if num_outliers > 0:
        # Menampilkan semua nilai outlier yang unik untuk kolom tersebut
        unique_outliers = sorted(outliers[col].unique())
        print(f"Nilai Outlier Unik untuk '{col}': {unique_outliers}")
    
    print("-"*50)
    
    # 5. Visualisasi dengan Boxplot
    plt.figure(figsize=(6, 4))
    sns.boxplot(x=df[col])
    
    # Tambahkan garis bantu untuk lower_bound dan upper_bound (opsional)
    plt.axvline(x=lower_bound, color='red', linestyle='--', label='Lower Bound')
    plt.axvline(x=upper_bound, color='blue', linestyle='--', label='Upper Bound')
    
    plt.title(f'Boxplot untuk Kolom {col}')
    plt.legend()
    plt.show()
```

**Important Point** : Berdasarkan apa yang sudah kita deteksi mengenai outliers, maka kita akan mengambil langkah dengan menghapus outliers dengan nilai yang ekstrem saja. Berikut yang akan kita lakukan :

#### 3.5 Analisis Kolerasi Numerik

Melakukan pengecekan korelasi dengan metode spearman karena data kita tidak terdistribusi normal

- **Mengapa Menggunakan Metode Spearman:**  
  - Spearman adalah metode non-parametrik yang mengukur korelasi berdasarkan peringkat (ranks), sehingga tidak terpengaruh oleh asumsi normalitas data.
  - Metode ini mampu mendeteksi hubungan monotonic antara variabel, meskipun hubungan tersebut tidak linier, sehingga lebih cocok untuk dataset yang tidak terdistribusi normal.

- **Mengapa Melihat Korelasi Antar Fitur Numerikal:**  
  - Analisis korelasi membantu mengidentifikasi hubungan antara fitur, sehingga kita bisa memahami apakah terdapat fitur yang sangat berkorelasi dan berpotensi redundan.
  - Mengetahui korelasi antar fitur juga berguna untuk mendeteksi multikolinearitas, yang dapat mempengaruhi stabilitas dan interpretabilitas model prediktif.


```python
# Daftar kolom yang ingin dilihat korelasinya
cols = ['Year', 'Engine_Size', 'Mileage', 'Price']

# 1) Membuat subset DataFrame hanya dengan 4 kolom numerik tersebut
df_corr = df[cols].copy()

# 2) Menghitung korelasi dengan metode Spearman (karena data non-normal)
corr_matrix = df_corr.corr(method='spearman')

# 3) Menampilkan nilai korelasi dalam bentuk tabel
print("Matriks Korelasi (Spearman):")
print(corr_matrix)

# 4) Visualisasi dengan Heatmap Seaborn
plt.figure(figsize=(6, 5))
sns.heatmap(corr_matrix, 
            annot=True,        # menampilkan nilai korelasi di tiap sel
            cmap='coolwarm',   # skema warna
            vmin=-1, vmax=1,   # mengatur rentang warna dari -1 s/d 1
            square=True)       # sel heatmap berbentuk kotak

plt.title('Spearman Correlation Heatmap')
```

####  Interpretasi Hasil Heatmap Korelasi:

- `Engine_Size` juga memiliki korelasi **positif** dengan `Price` sebesar **0.36**, artinya semakin besar ukuran mesin, cenderung semakin mahal mobil tersebut.
- Sebaliknya, `Mileage` berkorelasi **negatif** dengan `Price` sebesar **-0.25**, yang menunjukkan bahwa semakin jauh mobil telah digunakan, semakin rendah harga jualnya.

### Uji Statistik ANOVA pada Fitur Kategorikal terhadap Harga

Untuk melengkapi analisis visual sebelumnya, kita melakukan **Uji ANOVA (Analysis of Variance)** terhadap fitur-fitur kategorikal terhadap variabel target `Price`. Uji ini digunakan untuk mengetahui apakah **rata-rata harga antar kategori berbeda secara signifikan secara statistik**.

Kenapa menggunakan ANOVA?
- **ANOVA (Analysis of Variance)** cocok untuk mengetahui apakah setidaknya ada satu kategori dalam fitur kategorikal yang memiliki harga berbeda secara signifikan dibanding yang lain.
- Kita sebelumnya telah mengetahui dari visualisasi bahwa terdapat variasi harga antar kategori, tetapi tanpa validasi statistik, kita tidak bisa menyimpulkan bahwa perbedaan tersebut tidak disebabkan oleh variasi acak.
- Dengan ANOVA, kita mendapatkan **F-value** (seberapa besar variasi antar kelompok dibanding dalam kelompok) dan **p-value** (probabilitas bahwa perbedaan ini hanya karena kebetulan).

Jika **p-value < 0.05**, maka kita menyimpulkan bahwa perbedaan harga antar kategori signifikan dan fitur tersebut relevan secara statistik terhadap prediksi harga.

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Ambil semua fitur kategorikal
categorical_features = df.select_dtypes(include='object').columns.tolist()

# Inisialisasi list hasil
anova_results = []
print("Signifikan bila p-value < 0.05")
print("Semakin besar F-value, semakin penting fitur tersebut")
# Loop untuk setiap fitur kategorikal
for col in categorical_features:
    model = ols(f'Price ~ C({col})', data=df).fit()
    anova = sm.stats.anova_lm(model, typ=2)
    f_val = anova['F'][0]
    p_val = anova['PR(>F)'][0]
    
    if p_val < 0.01:
        level = "Sangat Signifikan"
    elif p_val < 0.05:
        level = "Signifikan"
    elif p_val < 0.1:
        level = "Kurang Signifikan"
    else:
        level = "Tidak Signifikan"

    anova_results.append({
        'Feature': col,
        'F-value': round(f_val, 2),
        'p-value': p_val,
        'Significance': level
    })

# Tampilkan hasil sebagai DataFrame
anova_df = pd.DataFrame(anova_results).sort_values(by='F-value', ascending=False).reset_index(drop=True)
display(anova_df)
```

```python
# Buat barplot F-value untuk visualisasi pengaruh fitur
plt.figure(figsize=(10, 6))
sns.barplot(data=anova_df, x='F-value', y='Feature', palette='viridis')
plt.title('F-value dari ANOVA per Fitur Kategorikal terhadap Harga')

plt.xlabel('F-value')
plt.ylabel('Fitur Kategorikal')
plt.tight_layout()
plt.show()
```

Uji statistik ANOVA yang dilakukan terhadap seluruh fitur kategorikal ini hasilnya menunjukkan bahwa seluruh fitur, termasuk `Color` dan `Region`, memiliki p-value < 0.05 yang berarti secara statistik **signifikan terhadap perbedaan harga**. Namun, fitur `Region` memiliki nilai F yang paling rendah (sekitar 4.26) di antara fitur lainnya. Hal ini menunjukkan bahwa meskipun perbedaannya signifikan secara statistik, **besarnya pengaruh atau kontribusi fitur `Region` terhadap variasi harga sangat kecil**, dan informasinya sudah cukup terwakili oleh fitur `Origin`. Selain itu, `Region` memiliki banyak kategori dan dapat menambah kompleksitas pada model ketika dilakukan encoding. Oleh karena itu, fitur `Region` akan dihapus dalam tahap preprocessing untuk menjaga kesederhanaan model.

**Insight dan Rekomendasi**:
Fitur-fitur seperti `Gear_Type`, `Options`, `Fuel_Type`, `Origin`, dan `Type` terbukti memiliki hubungan yang kuat dan signifikan secara statistik terhadap harga. Oleh karena itu, fitur-fitur ini disarankan untuk digunakan dalam tahap modeling prediktif harga mobil bekas. Sebaliknya, meskipun `Region` signifikan secara statistik, pengaruhnya terhadap harga sangat kecil secara praktikal dan bersifat redundan dengan `Origin`, sehingga fitur ini akan dihapus untuk menyederhanakan model dan menghindari duplikasi informasi.

```python
#drop engine size
df_clean.drop(columns=['Engine_Size'], inplace=True)
```

```python
import category_encoders as ce

# Kolom yang akan diencode
loo_encoded_feature = ['Type', 'Make', 'Region', 'Color']  # Hanya Type yang di-target encode  
# binary_features = []  # Sisanya tetap pakai Binary 
ordinal_features = ['Gear_Type', 'Origin', 'Options', 'Fuel_Type']

# Target encode hanya pada kolom 'Type'
loo_encoder = ce.LeaveOneOutEncoder(cols=loo_encoded_feature)
df_encoded = loo_encoder.fit_transform(df, df['Price'])  # penting: gunakan target saat fit

# Binary encoding untuk fitur lainnya
# binary_encoder = ce.BinaryEncoder(cols=binary_features)
# df_encoded = binary_encoder.fit_transform(df_encoded)

# Ordinal encoding
ordinal_encoder = ce.OrdinalEncoder(cols=ordinal_features)
df_encoded = ordinal_encoder.fit_transform(df_encoded)
```

```python
from sklearn.preprocessing import StandardScaler

# Misalnya, df_encoded sudah mengandung semua kolom yang di-encoding
# Misalnya, kolom kontinu yang ingin di-scale adalah 'Mileage', 'Year', dan 'Engine_Size'
numerical_features = ['Mileage', 'Year']

scaler = StandardScaler()

# Terapkan scaling hanya pada kolom numerik kontinu
df_encoded[numerical_features] = scaler.fit_transform(df_encoded[numerical_features])

df_encoded.head()
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error, mean_absolute_percentage_error

# Model yang digunakan
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from xgboost import XGBRegressor

# Pisahkan fitur (X) dan target (y)
X = df_encoded.drop('Price', axis=1)
y = df_encoded['Price']

# Bagi data menjadi training dan testing set
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Daftar model
models = {
    'Linear Regression': LinearRegression(),
    'Ridge Regression': Ridge(alpha=1.0),
    'Lasso Regression': Lasso(alpha=0.1),
    'Elastic Net': ElasticNet(alpha=1.0, l1_ratio=0.5, random_state=42),
    'KNN Regressor': KNeighborsRegressor(n_neighbors=5),
    'Decision Tree Regressor': DecisionTreeRegressor(random_state=42),
    'Random Forest Regressor': RandomForestRegressor(n_estimators=100, random_state=42),
    'Gradient Boosting Regressor': GradientBoostingRegressor(random_state=42),
    'Support Vector Regressor': SVR(),
    'XGBoost Regressor': XGBRegressor(n_estimators=100, random_state=42, verbosity=0)
}

# Simpan hasil evaluasi semua model
results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    r2 = r2_score(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_test, y_pred)
    mape = mean_absolute_percentage_error(y_test, y_pred)
    
    results.append({
        'Model': name,
        'RMSE': rmse,
        'MAE': mae,
        'MAPE': mape,
        'R2': r2
    })

# Buat DataFrame hasil evaluasi
results_df = pd.DataFrame(results).set_index('Model')
results_df = results_df.sort_values(by='R2', ascending=False)  # Urutkan berdasarkan performa R2
results_df
```

```python
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBRegressor

# Parameter grid untuk RandomizedSearchCV
param_dist = {
    'n_estimators': [100, 200, 300, 400, 500],
    'max_depth': [3, 4, 5, 6, 7],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'subsample': [0.6, 0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.6, 0.7, 0.8, 0.9, 1.0],
    'gamma' : [0, 0.1, 0.2, 0.3],
    'min_child_weight' : [1, 3, 5]
}

# Inisialisasi model
xgb_model = XGBRegressor(objective="reg:squarederror", random_state=42)

# Setup RandomizedSearchCV
random_search = RandomizedSearchCV(
    estimator=xgb_model,
    param_distributions=param_dist,
    n_iter=50,
    scoring='r2',
    cv=5,
    verbose=1,
    random_state=42,
    n_jobs=-1
)

# Fitting pada training data
random_search.fit(X_train, y_train)

# Ambil model terbaik
best_xgb = random_search.best_estimator_

# Prediksi pada data test
y_pred = best_xgb.predict(X_test)

# Evaluasi performa
r2 = r2_score(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_pred)
mape = mean_absolute_percentage_error(y_test, y_pred)

# Tambahkan hasil ke results_df
results_df.loc['Tuned XGBoost Regressor (Random Search)'] = [rmse, mae, mape, r2]

# Tampilkan parameter dan skor terbaik
print("Best parameters found:", random_search.best_params_)
print("Best cross-validation R2 score:", random_search.best_score_)

# Tampilkan keseluruhan hasil
results_df = results_df.sort_values(by='R2', ascending=False)  # Urutkan berdasarkan performa R2
results_df
```
